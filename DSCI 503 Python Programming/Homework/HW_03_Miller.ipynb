## Problem 1: Dot Product

```python
def dot_product(x, y):
    # validate that the vectors are the same size
    if len(x) != len(y):
        return 0
    
    dot_product = 0
    for i in range(len(x)) :
        dot_product += x[i] * y[i]
    
    return dot_product
```

```python
v1 = [38, 9, 40, 34, 20, 16, 42, 36, 12, 1, 23, 46, 31, 19, 30, 33, 16, 43, 24, 41]
v2 = [43, 13, 35, 14, 26, 3, 36, 15, 42, 44, 45, 20, 17, 6, 47, 40, 38, 41, 31, 24]

print(f"The dot product of v1 and v2 is: {dot_product(v1, v2)}.")
```

## Problem 2: Amortization

```python
def count_payments(amount, rate, pmt):
    balance = amount
    number_of_payments = 0

    while balance > 0:
        balance = round(balance * (1 + rate) - pmt, 2) # add interest and subtract the payment
        number_of_payments += 1

    return number_of_payments
```

```python
amount = 160000
rate = 0.004

pmt = 850
increment = 50

while pmt <= 1050:
    number_of_payments = count_payments(amount, rate, pmt)
    print(f"{number_of_payments} monthly payments of ${pmt} would be required.")
    pmt += increment
```

## Problem 3: Minimum

```python
def minimum(x):
    smallest = x[0]
    for i in range(1, len(x)):
        if x[i] < smallest:
            smallest = x[i]
    return smallest
```

```python
list1 = [9.8, 7.4, 5.6, 4.8, 4.8, 5.3, 4.1, 9.6, 5.4]
list2 = [3.4, 7.6, 8.7, 7.5, 9.8, 7.5, 6.7, 8.7, 8.4]
list3 = ['St. Louis', 'Kansas City', 'Chicago', 'Little Rock', 'Omaha']

print(minimum(list1))
print(minimum(list2))
print(minimum(list3))
```

## Problem 4: Argmin

```python
def argmin(x):
    index_of_smallest = 0
    smallest = x[0]
    for i in range(1, len(x)):
        if x[i] < smallest:
            smallest = x[i]
            index_of_smallest = i
    
    return index_of_smallest
```

```python
print(argmin(list1))
print(argmin(list2))
print(argmin(list3))
```

## Problem 5: Find Smallest Elements

```python
def find_smallest(x, n = None):
    if n is None:
        return min(x) # return the smallest element
    elif n > len(x) or n <= 0:
        return x # return the whole list
    else:
        sorted_x = sorted(x)
        return sorted_x[:n] # return the n smallest elements
```

```python
my_list = [39, 74, 28, 64, 17, 28, 54, 53]

print(find_smallest(my_list))
print(find_smallest(my_list, 1))
print(find_smallest(my_list, 2))
print(find_smallest(my_list, 5))
print(find_smallest(my_list, 12))
```

## Problem 6: Find Unique Elements

```python
def unique(x):
    unique_list = []
    x.sort()
    for item in x:
        if item not in unique_list:
            unique_list.append(item)
    
    return unique_list
```

```python
int_list = [23, 16, 23, 12, 14, 23, 12, 19, 19]
str_list = ['cat', 'dog', 'dog', 'cat', 'bat', 'frog', 'dog', 'frog']

print(unique(int_list))
print(unique(str_list))
```

## Problem 7: Frequency Distribution

```python
def freq_dist(x):
    values = unique(x)
    counts = []

    for value in values:
        counts.append(x.count(value))

    return values, counts
```

```python
grades = ['A', 'D', 'A', 'C', 'B', 'F', 'A', 'D', 'C',
          'B', 'F', 'A', 'C', 'B', 'A', 'B', 'B', 'C',
          'B', 'F', 'D', 'D', 'A', 'C', 'B', 'B', 'D']

values, counts = freq_dist(grades)

print("Unique Values:", values)
print("Frequencies:  ", counts)
```

## Problem 8: Weighed Means

```python
def mean(x, w = None):
    if w is None:
        # standard mean
        return sum(x) / len(x)
    else:
        # weighted mean
        weighted_sum = 0
        total_weight = 0

        for i in range(len(x)):
            weighted_sum += x[i] * w[i]
            total_weight += w[i]

        return weighted_sum / total_weight
```

```python
values = [4, 7, 3, 5, 2, 6, 8, 2, 4, 8]
weights = [2, 1, 3, 1, 2, 3, 1, 4, 2, 1]

standard_mean = mean(values)
weighted_mean = mean(values, weights)

print("Standard Mean:", standard_mean)
print("Weighted Mean:", weighted_mean)
```
