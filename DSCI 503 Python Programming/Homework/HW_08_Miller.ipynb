```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, classification_report
from sklearn.preprocessing import OneHotEncoder
```

## Problem 1: Diamonds Dataset

```python
diamonds = pd.read_csv('diamonds.txt', sep='\t')
# add columns for the natural log of carat and price
diamonds.loc[:, 'ln_carat'] = np.log(diamonds.carat)
diamonds.loc[:, 'ln_price'] = np.log(diamonds.price)
diamonds.head(5)
```

```python
X1_num = diamonds.ln_carat.values.reshape(-1, 1)
X1_cat = diamonds.loc[:, ['cut', 'color', 'clarity']].values
y1 = diamonds.ln_price.values

print(f"Numerical Feature Array Shape:   {X1_num.shape}")
print(f"Categorical Feature Array Shape: {X1_cat.shape}")
print(f"Label Array Shape:               {y1.shape}")
```

```python
encoder = OneHotEncoder(sparse_output=False)
encoder.fit(X1_cat)
X1_enc = encoder.transform(X1_cat)
print(f"Encoded Feature Array Shape: {X1_enc.shape}")
```

```python
X1 = np.hstack((X1_num, X1_enc))
print(f"Feature Array Shape: {X1.shape}")
```

```python
X1_train, X1_hold, y1_train, y1_hold = train_test_split(X1, y1, test_size=0.2, random_state=1)
x1_valid, X1_test, y1_valid, y1_test = train_test_split(X1_hold, y1_hold, test_size=0.5, random_state=1)
print(f"Training Features Shape:   {X1_train.shape}")
print(f"Validation Features Shape: {x1_valid.shape}")
print(f"Test Features Shape:       {X1_test.shape}")
```

### Linear Regression Model with One Feature

```python
dia_mod = LinearRegression()
dia_mod.fit(X1_train[:, 0:1], y1_train)

dia_train_r2 = dia_mod.score(X1_train[:, 0:1], y1_train)
dia_valid_r2 = dia_mod.score(x1_valid[:, 0:1], y1_valid)

print(f"Training r-Squared:   {np.round(dia_train_r2, 4)}")
print(f"Validation r-Squared: {np.round(dia_valid_r2, 4)}")
```

### Linear Regression Model with Several Features

```python
dia_mod2 = LinearRegression()
dia_mod2.fit(X1_train, y1_train)
dia2_train_r2 = dia_mod2.score(X1_train, y1_train)
dia2_valid_r2 = dia_mod2.score(x1_valid, y1_valid)
print(f"Training r-Squared:   {np.round(dia2_train_r2, 4)}")
print(f"Validation r-Squared: {np.round(dia2_valid_r2, 4)}")
```

```python
print(f"Testing r-Squared:  {np.round(dia_mod2.score(X1_test, y1_test), 4)}")
```

## Problem 2: Census Dataset

```python
census = pd.read_csv('census.txt', sep='\t')
census.head(10)
```

```python
print(census.shape)
```

```python
print(census.salary.value_counts())
```

### Prepare the Data

```python
X2_num = census.loc[:, ['age', 'fnlwgt', 'educ_num', 'capital_gain',
                        'capital_loss', 'hrs_per_week']].values

X2_cat = census.loc[:, ['workclass', 'education', 'marital_status', 
                        'occupation', 'relationship', 'race', 'sex', 
                        'native_country']].values

y2 = census.salary.values

print(f"Numerical Feature Array Shape:   {X2_num.shape}")
print(f"Categorical Feature Array Shape: {X2_cat.shape}")
print(f"Label Array Shape:               {y2.shape}")
```

```python
encoder2 = OneHotEncoder(sparse_output=False)
encoder2.fit(X2_cat)
X2_enc = encoder2.transform(X2_cat)
print(f"Encoded Feature Array Shape: {X2_enc.shape}")
```

```python
X2 = np.hstack((X2_num, X2_enc))
print(f"Feature Array Shape: {X2.shape}")
```

```python
X2_train, X2_hold, y2_train, y2_hold = train_test_split(X2, y2, test_size=0.3, random_state=1, stratify=y2)
X2_valid, X2_test, y2_valid, y2_test = train_test_split(X2_hold, y2_hold, test_size=0.5, random_state=1, stratify=y2_hold)
print(f"Training Features Shape:   {X2_train.shape}")
print(f"Validation Features Shape: {X2_valid.shape}")
print(f"Test Features Shape:       {X2_test.shape}")
```

### Logistic Regression Model

```python
lr_mod = LogisticRegression(solver='lbfgs', max_iter=1000, C=10e1000)
lr_mod.fit(X2_train, y2_train)
print(f"Training Accuracy:   {lr_mod.score(X2_train, y2_train)}")
print(f"Validation Accuracy: {lr_mod.score(X2_valid, y2_valid)}")
```

### Decision Tree Models

```python
dt_train_acc = []
dt_valid_acc = []
depth_range = range(2, 31)

for depth in depth_range:
    np.random.seed(1)
    temp_tree = DecisionTreeClassifier(max_depth=depth)
    temp_tree.fit(X2_train, y2_train)
    dt_train_acc.append(temp_tree.score(X2_train, y2_train))
    dt_valid_acc.append(temp_tree.score(X2_valid, y2_valid))

dt_idx = np.argmax(dt_valid_acc)
dt_opt_depth = depth_range[dt_idx]
dc_opt_train_acc = dt_train_acc[dt_idx]
dc_opt_valid_acc = dt_valid_acc[dt_idx]
print(f"Optimal value for max_depth:           {dt_opt_depth}")
print(f"Training Accuracy for Optimal Model:   {dc_opt_train_acc}")
print(f"Validation Accuracy for Optimal Model: {dc_opt_valid_acc}")
```

```python
plt.figure(figsize=(10, 6))
plt.plot(depth_range, dt_train_acc, label='Training', color='blue')
plt.plot(depth_range, dt_valid_acc, label='Validation', color='orange')
plt.xlabel("Max Depth")
plt.ylabel("Accuracy")
plt.legend()
plt.show()
```

### Random Forest Models

```python
rf_train_acc = []
rf_valid_acc = []

for depth in depth_range:
    np.random.seed(1)
    temp_forest = RandomForestClassifier(n_estimators=100, max_depth=depth, random_state=1)
    temp_forest.fit(X2_train, y2_train)
    rf_train_acc.append(temp_forest.score(X2_train, y2_train))
    rf_valid_acc.append(temp_forest.score(X2_valid, y2_valid))

rf_idx = np.argmax(rf_valid_acc)
rf_opt_depth = depth_range[rf_idx]
rf_opt_train_acc = rf_train_acc[rf_idx]
rf_opt_valid_acc = rf_valid_acc[rf_idx]
print(f"Optimal value for max_depth:           {rf_opt_depth}")
print(f"Training Accuracy for Optimal Model:   {rf_opt_train_acc}")
print(f"Validation Accuracy for Optimal Model: {rf_opt_valid_acc}")
```

```python
plt.figure(figsize=(10, 6))
plt.plot(depth_range, rf_train_acc, label='Training', color='blue')
plt.plot(depth_range, rf_valid_acc, label='Validation', color='orange')
plt.xlabel("Max Depth")
plt.ylabel("Accuracy")
plt.legend()
plt.show()
```

### Evaluate Final Model

```python
# The model with the highest validation accuracy is the RandomForestClassifier with depth 19
np.random.seed(1)
final_model = RandomForestClassifier(n_estimators=100, max_depth=19, random_state=1)
final_model.fit(X2_train, y2_train)

print(f"Training Accuracy for Final Model:   {np.round(final_model.score(X2_train, y2_train))}")
print(f"Validation Accuracy for Final Model: {np.round(final_model.score(X2_valid, y2_valid))}")
print(f"Testing Accuracy for Final Model:    {np.round(final_model.score(X2_test, y2_test))}")
```

```python
test_pred = final_model.predict(X2_test)
cm = confusion_matrix(y2_test, test_pred)
cm_df = pd.DataFrame(cm, index=['<=50K', '>50k'], columns=['<=50K', '>50k'])
cm_df
```

```python
print(classification_report(y2_test, test_pred))
```
