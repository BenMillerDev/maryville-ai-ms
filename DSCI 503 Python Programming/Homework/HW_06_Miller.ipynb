```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Problem 1: Iris Dataset

```python
iris_means = iris.loc[:,:].groupby('species').mean()
iris_means
```

## Problem 2: Ames Housing - Neighborhoods

```python
ames.head(8)
```

```python
print(ames.columns.values)
```

```python
# select the Neighborhood and SalePrice columns, 
# group by Neighborhood and select the SalePrice, 
# find the mean and sort
ames_nbhd = ames.loc[:, ['Neighborhood', 'SalePrice']].\
            groupby('Neighborhood')['SalePrice'].\
            mean().sort_values(ascending=True)

ames_nbhd.head(5)
```

```python
plt.figure(figsize=(6, 10))
plt.barh(ames_nbhd.index, ames_nbhd.values, color='purple')
plt.xlabel('Mean Sale Price')
plt.ylabel('Neighborhood')
plt.title('Mean Sale Price by Neighborhood')
plt.show()
```

## Problem 3: Ames Housing – House Styles

```python
# groupby House Style, select the Sale Price, and count
# reset_index() converts the Series result back into a DataFrame
ames_style = ames.loc[:, ['House Style', 'SalePrice']].\
               groupby('House Style')['SalePrice'].\
               count().sort_values(ascending=True).reset_index()

ames_style.rename(columns={'SalePrice':'Count'}, inplace=True)
ames_style
```

```python
plt.figure(figsize=(6, 4))
plt.bar(ames_style['House Style'], ames_style['Count'], color='orange')
plt.xlabel('House Style')
plt.ylabel('Count')
plt.title('Houses Sold by Style')
plt.show()
```

## Problem 7: Batting Data – Batting Averages

```python
batting_avg = batting.loc[:, ['teamID', 'yearID', 'H', 'AB']].\
                groupby(['teamID', 'yearID']).sum()

batting_avg.loc[:, 'BA'] = batting_avg.loc[:, 'H'] / batting_avg.loc[:, 'AB']
batting_avg = batting_avg.sort_values(by='BA', ascending=False)
batting_avg.head(10)
```

## Problem 8: Batting Data – Cardinals vs. Cubs

```python
stl_batting = batting.loc[(batting.teamID == 'SLN') & (batting.yearID >= 1900), ['yearID', 'H', 'AB']].\
                groupby('yearID').sum()

stl_batting.loc[:, 'BA'] = stl_batting.loc[:, 'H'] / stl_batting.loc[:, 'AB']

chi_batting = batting.loc[(batting.teamID == 'CHN') & (batting.yearID >= 1900), ['yearID', 'H', 'AB']].\
                groupby('yearID').sum()
chi_batting.loc[:, 'BA'] = chi_batting.loc[:, 'H'] / chi_batting.loc[:, 'AB']
```

```python
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(stl_batting.index, stl_batting['BA'], label='Cardinals', color='red')
plt.plot(chi_batting.index, chi_batting['BA'], label='Cubs', color='blue')
plt.xlabel('Year')
plt.ylabel('Batting Average')
plt.title('Batting Average by Year')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(stl_batting.index, stl_batting['H'], label='Cardinals', color='red')
plt.plot(chi_batting.index, chi_batting['H'], label='Cubs', color='blue')
plt.xlabel('Year')
plt.ylabel('Home Runs')
plt.title('Home Runs by Year')
plt.legend()

plt.show()
```

```python
cardinals_ba_proportion = np.mean(stl_batting.BA > chi_batting.BA)
print(round(cardinals_ba_proportion, 4))
```

```python
cubs_ba_proportion = np.mean(chi_batting.BA > stl_batting.BA)
print(round(cubs_ba_proportion, 4))
```
