## Problem 1: Altering Lists

```python
employees = ["Beth", "Fred", "Drew", "Alex", "Emma", "Chad"]
years = [15, 11, 8, 5, 4, 2]
```

```python
# Insert a new employee, Gina, with 12 years of service
employees.insert(1, "Gina")
years.insert(1, 12)

# Insert a new employee, Herb, with 3 years of service
employees.insert(6, "Herb")
years.insert(6, 3)

# Insert a new employee, Iris, with 0 years of service
employees.append("Iris")
years.append(0)

print(employees)
print(years)
```

```python
# Remove Fred and his years of service
fred_index = employees.index("Fred")
del employees[fred_index]
del years[fred_index]

# Correct Emma's years of service from 4 to 6, while maintaining the descending order
emma_index = employees.index("Emma")
years[emma_index] = 6 # update Emma's years of service

alex_index = emma_index - 1 # Alex is right before Emma in the list

# use tuples to swap the values of Emma and Alex
# alex, emma = emma, alex
employees[alex_index], employees[emma_index] = employees[emma_index], employees[alex_index]
years[alex_index], years[emma_index] = years[emma_index], years[alex_index]

print(employees)
print(years)
```

## Problem 3: Calculating Sum of Squared Errors

```python
length = [22.7, 22.4, 25.8, 21.3, 20.1, 22.1, 21.1, 25.3, 26.9, 26.9,
23.0, 23.8, 26.2, 20.4, 23.0, 21.9, 23.5, 27.8, 25.3, 25.9]
weight = [9.2, 8.8, 10.7, 8.3, 6.2, 8.6, 7.2, 11.2, 10.5, 11.3,
9.6, 9.9, 10.9, 5.9, 9.5, 9.1, 9.7, 11.6, 10.2, 10.5]
```

```python
pred_weight = []
for marmot_length in length:
    estimated_weight = round(0.6 * marmot_length - 5.2, 2)
    pred_weight.append(estimated_weight)

print (pred_weight)
```

```python
sum_of_squared_difference = 0
for i in range(0, len(pred_weight)):
    difference =  weight[i] - pred_weight[i]
    squared_difference = difference ** 2
    sum_of_squared_difference += squared_difference

print(round(sum_of_squared_difference, 4))
```

## Problem 5: Determining Monthly Payments

```python
i = 0.004
n = 30 * 12
start = 100000
end = 200000
increment = 10000

iterations = int((end - start) / increment) + 1 # +1 to include the end value

# L = PMT * (1 - (1 + i)^-n) / i
# I will call the right side the "fraction"
fraction = (1 - (1 + i) ** -n) / i

#PMT = L / fraction
for index in range(0, iterations):   
  # add the increment to the start value, starting from 100,000 and increasing by 10,000 each iteration
  loan_amount = start + index * increment
  monthly_payment = loan_amount / fraction
  print(f"A loan of ${round(loan_amount,2)} would require monthly payments of {round(monthly_payment, 2)}.")
```

## Problem 7: Calculating Price of a Bulk Order

```python
quantities = [84, 100, 126, 150, 186, 200, 216, 248]
base_price = 350.00
price_over_100 = 300.00
price_over_200 = 250.00

for quantity in quantities:  

  if quantity <= 100:
    price = base_price * quantity
  elif quantity <= 200:
    discounted_units_over_100 = quantity - 100
    price = (base_price * 100) + (price_over_100 * discounted_units_over_100)
  else:
    discounted_units_over_200 = quantity - 200
    price = (base_price * 100) + (price_over_100 * 100) + (price_over_200 * discounted_units_over_200)

  print(f"The cost for an order of {quantity} widgets is ${round(price, 2)}.")
```
