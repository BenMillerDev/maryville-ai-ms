## Problem 1: Arithmetic and Geometric Means

```python
x0 = 11
x1 = 12
x2 = 17
x3 = 19

total = x0 + x1 + x2 + x3
product = x0 * x1 * x2 * x3

arith_mean = total / 4
geom_mean = product ** (1/4)

print("Sum:            ", total)
print("Product:        ", product)
print("Arithmetic Mean:", round(arith_mean,2))
print("Geometric Mean: ", round(geom_mean,2))
```

## Problem 2: Calculating Bill

```python
salad = 3.95
burger = 8.95
soda = 2.50

ss_cost = salad * 2
hb_cost = burger * 2
ds_cost = soda * 2

sales_tax_percent = 0.0475
tip_percent = 0.15

subtotal = round(ss_cost + hb_cost + ds_cost, 2)
tax = round(subtotal * sales_tax_percent, 2)
tip = round(subtotal * tip_percent, 2)
total_charge = round(subtotal + tax + tip, 2)

print("2 Side Salads:    $", ss_cost, sep="")
print("2 Hamburgers:    $", hb_cost, sep="")
print("2 Diet Sodas:     $", ds_cost, sep="")
print("-" * 22)
print("Subtotal:        $", subtotal, sep="")
print("Tax (4.75%):     $", tax, sep="")
print("Tip (15%):       $", tip, sep="")
print("-" * 22)
print("Total Charge:   $", total_charge, sep="")
```

## Problem 3: Volume of a Sphere

```python
pi = 3.14159
r1 = 4.6
r2 = 7.2
r3 = 9.7

v1 = (4/3) * pi * (r1 ** 3)
v2 = (4/3) * pi * (r2 ** 3)
v3 = (4/3) * pi * (r3 ** 3)

print(f"The volume of a sphere with radius {r1} is equal to {round(v1, 3)}.")
print(f"The volume of a sphere with radius {r2} is equal to {round(v2, 3)}.")
print(f"The volume of a sphere with radius {r3} is equal to {round(v3, 3)}.")
```

## Problem 7: Probability

```python
prob_A = 0.6
prob_B = 0.3

prob_A_and_B = prob_A * prob_B
prob_A_or_B = prob_A + prob_B - prob_A_and_B

print("Probability that both people are late:       ", prob_A_and_B)
print("Probability that at least one person is late:", prob_A_or_B)
```
