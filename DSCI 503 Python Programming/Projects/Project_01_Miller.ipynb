# Introduction
This project involves reading and manipulating data from a text file.  The text file contains socioeconomic data for 184 countries.

```python
import pandas as pd
df = pd.read_csv('gapminder_data.txt', sep='\t')
df.head(10)
```

```python
country = df.country.to_list()
continent = df.continent.to_list()
year = df.year.to_list()
population = df.population.to_list()
life_exp = df.life_exp.to_list()
pcgdp = df.gdp_per_cap.to_list()
gini = df.gini.to_list()
```

```python
N = 100
print('Country: ', country[N])
print('Continent: ', continent[N])
print('Year: ', year[N])
print('Population: ', population[N])
print('Life Expectancy: ', life_exp[N])
print('Per Capita GDP: ', pcgdp[N])
print('Gini Index: ', gini[N])
```

# Part 1: Displaying Past 20 Years of US Data
This section will involve searching through the lists and finding data related to the United States in the last 20 years.

```python
# Align the output so there are 4 spaces between the values in each column
# Explanation of the formatting: Length of the word + 4 spaces, with left alignment
# "Year" (4) + 4 spaces = 8
# The country name "United States" (13) + 4 spaces = 17
# "Population" (10) + 4 spaces = 14
# "LExp" (4) + 4 spaces = 8
# "pcGDP" (5) + 4 spaces = 9
# "Gini" is at the end, so no extra spaces needed
header = f"{"Year":<8}{"Country":<17}{"Population":<14}{"LExp":<8}{"pcGDP":<9}{"Gini"}"
print(header)
print("-" * len(header))

# Loop through the data and print US data for the past 20 years
for i in range(len(country)):
    if country[i] == 'United States' and year[i] >= 1999:
        print(f"{year[i]:<8}{country[i]:<17}{population[i]:<14}{life_exp[i]:<8}{pcgdp[i]:<9}{gini[i]}")
```

# Part 2: Selecting the 2018 Data
In this section I will create lists containing data from 2018 for each country.

```python
country_2018 = []
population_2018 = []
continent_2018 = []
life_exp_2018 = []
pcgdp_2018 = []
gini_2018 = []

for i in range(len(country)):
    if year[i] == 2018:
        country_2018.append(country[i])
        population_2018.append(population[i])
        continent_2018.append(continent[i])
        life_exp_2018.append(life_exp[i])
        pcgdp_2018.append(pcgdp[i])
        gini_2018.append(gini[i])
```

```python
global_population_2018 = sum(population_2018)
print(f"The global population in 2018 was {global_population_2018}.")
```

# Part 3: Identifying Countries with Largest and Smallest Populations
In this section I will display countries with the largest and smallest populations in 2018

```python
sorted_population_2018 = sorted(population_2018)
tenth_largest_population = sorted_population_2018[-10] # the 10th largest population

print("Countries with Largest Populations in 2018")
print("------------------------------------------------------")

for i in range(len(country_2018)):
    if population_2018[i] >= tenth_largest_population:
        print(f"The population of {country_2018[i]} in 2018 was {population_2018[i]}.")
```

```python
tenth_smallest_population = sorted_population_2018[9] # the 10th smallest population

print("Countries with Smallest Populations in 2018")
print("------------------------------------------------------")

for i in range(len(country_2018)):
    if population_2018[i] <= tenth_smallest_population:
        print(f"The population of {country_2018[i]} in 2018 was {population_2018[i]}.")
```

# Part 4: Identifying Countries with Highest and Lowest Life Expectancies
In this section I will display data for countries with the highest and lowest life expectancies in 2018.

```python
print("Countries with Highest Life Expectancy in 2018")
print("------------------------------------------------------")

sorted_life_exp_2018 = sorted(life_exp_2018)
tenth_largest_life_exp = sorted_life_exp_2018[-10] # the 10th largest life expectancy

for i in range(len(country_2018)):
    if life_exp_2018[i] >= tenth_largest_life_exp:
        print(f"The life expectancy of {country_2018[i]} in 2018 was {life_exp_2018[i]}.")
```

```python
tenth_lowest_life_exp = sorted_life_exp_2018[9] # the 10th smallest life expectancy

print("Countries with Lowest Life Expectancy in 2018")
print("------------------------------------------------------")

for i in range(len(country_2018)):
    if life_exp_2018[i] <= tenth_lowest_life_exp:
        print(f"The life expectancy of {country_2018[i]} in 2018 was {life_exp_2018[i]}.")
```

# Part 5: Calculating GDP by Country
in this section I will create a list that contains the GDP for each country, calculated by multiplying the per capita GDP by the population.

```python
gdp_2018 = []
for i in range(len(country_2018)):
    gdp = pcgdp_2018[i] * population_2018[i]
    gdp_2018.append(gdp)

print(f"The total global GDP in 2018 was ${sum(gdp_2018)}.")
```

```python
min_gdp_2018 = min(gdp_2018)
max_gdp_2018 = max(gdp_2018)

min_gdp_index = gdp_2018.index(min_gdp_2018)
max_gdp_index = gdp_2018.index(max_gdp_2018)

print(f"The country with the highest GDP in 2018 was {country_2018[max_gdp_index]} with a GDP of ${max_gdp_2018}.")
print(f"The country with the lowest GDP in 2018 was {country_2018[min_gdp_index]} with a GDP of ${min_gdp_2018}.")
```

# Part 6: Grouping by Continent
In this section I will group the data by continent and find the total population, per capita GDP, and life expectancy for each.

```python
continent_names = ["africa", "americas", "asia", "europe"]

pcgdp_2018_by_continent = []
pop_2018_by_continent = []
life_exp_2018_by_continent = []

for continent in continent_names:
    temp_life_exp = 0
    temp_pop = 0
    temp_gdp = 0

    for i in range(len(country_2018)):
        if continent_2018[i] == continent:  
            temp_pop += population_2018[i]
            temp_gdp += gdp_2018[i]
            temp_life_exp += life_exp_2018[i] * population_2018[i] # weighted sum
    
    temp_life_exp = temp_life_exp / temp_pop # find the average life expectancy
    temp_gdp = temp_gdp / temp_pop # find the average per capita GDP
    
    temp_gdp = round(temp_gdp) # round to the nearest whole number
    temp_life_exp = round(temp_life_exp, 1) # round to one decimal place

    pcgdp_2018_by_continent.append(temp_gdp)
    pop_2018_by_continent.append(temp_pop)
    life_exp_2018_by_continent.append(temp_life_exp)

# Explanation of formatting:
# "Continent" (9), left aligned
# "Population" (10) + 4 spaces = 14, right aligned
# "pcGDP" (5) + 4 spaces = 9, right aligned
# "Life Exp" (8) + 4 spaces = 12, right aligned
header = f"{'Continent':<9}{'Population':>14}{'pcGDP':>9}{'Life Exp':>12}"
print(header)
print("-" * len(header))
for i in range(len(continent_names)):
    formatted_name = f"{continent_names[i].capitalize():<9}"
    formatted_pop = f"{pop_2018_by_continent[i]:>14}"
    formatted_pcgdp = f"{pcgdp_2018_by_continent[i]:>9}"
    formatted_life_exp = f"{life_exp_2018_by_continent[i]:>12}"
    
    print(formatted_name + formatted_pop + formatted_pcgdp + formatted_life_exp)
```
