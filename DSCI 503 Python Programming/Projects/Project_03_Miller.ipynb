## Part 3: Working with Categorical Variables
create lists to specify the order of the three categorical variables

```python
clarity_levels = ['I1', 'SI2', 'SI1', 'VS2', 'VS1', 'VVS2', 'VVS1', 'IF']
cut_levels = ['Fair', 'Good', 'Very Good', 'Premium', 'Ideal']
color_levels = ['J', 'I', 'H', 'G', 'F', 'E', 'D']
```

use these rankings to tell Pandas the correct order for the categories

```python
diamonds.clarity = pd.Categorical(diamonds.clarity, categories=clarity_levels)
diamonds.cut = pd.Categorical(diamonds.cut, categories=cut_levels)
diamonds.color = pd.Categorical(diamonds.color, categories=color_levels)
```

create a list of colors to use as palettes for visualizations

```python
clarity_pal = ['red', 'orangered', 'orange', 'yellow', 'yellowgreen', 'green', 'aqua', 'blue']
cut_pal = ['red',  'orange', 'yellow', 'green', 'blue']
color_pal = ['red', 'orange', 'yellow', 'yellowgreen', 'green', 'aqua', 'blue']
```

## Part 5: Scatterplots of Price Against Carat
create scatter plots to examine the relationship between price and carat

```python
plt.figure(figsize=(8, 6))
plt.scatter(diamonds.carat, diamonds.price, s=20, alpha=0.4, color='blue')
plt.xlabel('Carat Size')
plt.ylabel('Price (in $)')
plt.title('Relationship between Diamond Price and Carat Size')
plt.show()
```

reproduce the scatter plot, but color the points by clarity

```python
plt.figure(figsize=(8, 6))
for level in clarity_levels:
    subset = diamonds[diamonds.clarity == level]
    clarity_color = clarity_pal[clarity_levels.index(level)]
    plt.scatter(subset.carat, subset.price, s=20, alpha=0.4, label=level, color=clarity_color)
    plt.legend()
    
plt.xlabel('Carat Size')
plt.ylabel('Price (in $)')
plt.title('Relationship between Diamond Price and Carat Size')
plt.show()
```

separate the points for each of the eight clarity levels into its own
subplot

```python
plt.figure(figsize=(12, 6))

for i in range(len(clarity_levels)):
    clarity = clarity_levels[i]
    plt.subplot(2, 4, i + 1)
    subset = diamonds[diamonds.clarity == clarity]
    plt.scatter(subset.carat, subset.price, s=20, alpha=0.4, color=clarity_pal[i])
    plt.title(clarity)
    plt.xlabel('Carat Size')
    plt.ylabel('Price (in $)')
    plt.xlim(0, 4)
    plt.ylim(0, 20000)

plt.tight_layout()
plt.show()
```

## Part 6: Applying Logarithmic Transformations
add two new columns to diamonds to store transformed variables

```python
diamonds.loc[:, 'ln_carat'] = np.log(diamonds.carat)
diamonds.loc[:, 'ln_price'] = np.log(diamonds.price)
```

create a histogram to view the distribution of diamond prices and the log of the diamond prices

```python
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.hist(diamonds.price, bins=20, color='violet', alpha=0.7, edgecolor='black')
plt.xlabel('Price')
plt.ylabel('Count')
plt.title('Histogram of Prices')

plt.subplot(1, 2, 2)
plt.hist(diamonds.ln_price, bins=20, color='violet', alpha=0.7, edgecolor='black')
plt.xlabel('Natural Log of Price')
plt.ylabel('Count')
plt.title('Histogram of Log-Prices')

plt.show()
```

create a histogram to view the distribution of diamond carat sizes and the log of the diamond carat sizes

```python
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.hist(diamonds.carat, bins=20, color='cyan', alpha=0.7, edgecolor='black')
plt.xlabel('Carat Size')
plt.ylabel('Count')
plt.title('Histogram of Carat Sizes')

plt.subplot(1, 2, 2)
plt.hist(diamonds.ln_carat, bins=20, color='cyan', alpha=0.7, edgecolor='black')
plt.xlabel('Natural Log of Carat Size')
plt.ylabel('Count')
plt.title('Histogram of Log-Carat Sizes')

plt.show()
```

## Part 7: Scatterplot of Transformed Variables
create a scatterplot of ln_price against ln_carat with points colored according to their clarity

```python
plt.figure(figsize=(8, 6))
for i in range(len(clarity_levels)):
    level = clarity_levels[i]
    subset = diamonds[diamonds.clarity == level]
    clarity_color = clarity_pal[i]
    plt.scatter(subset.ln_carat, subset.ln_price, s=20, alpha=0.4, label=level, color=clarity_color)
    plt.legend()

plt.xlabel('Natural Log of Carat Size')
plt.ylabel('Natural Log of Price')
plt.title('Relationship between Log-Price and Log-Carat Size')
plt.show()
```

## Part 8: Grouping by Cut
create a grouped DataFrame displaying the mean price and carat size for each cut level

```python
gb_cut = diamonds.loc[:, ['cut', 'price', 'carat']].groupby('cut', observed=False).mean()
gb_cut
```

create a bar chart to display the dataframe

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.bar(gb_cut.index, gb_cut.price, color=cut_pal, alpha=0.7, edgecolor='black')
plt.xlabel('Cut')
plt.ylabel('Mean Price')
plt.title('Mean Price')

plt.subplot(1, 2, 2)
plt.bar(gb_cut.index, gb_cut.carat, color=cut_pal, alpha=0.7, edgecolor='black')
plt.xlabel('Cut')
plt.ylabel('Mean Carat Size')
plt.title('Mean Carat Size by Cut')

plt.show()
```
