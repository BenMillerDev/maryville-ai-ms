# Problem 1 - Inverse of a Product

### Problem Background

In this problem, we will explore a property of matrix algebra. In particular, we will discuss how the inverse and matrix multiplication operations interact. It can be shown the following equation holds true for any two invertible matrices $A$ and $B$:


$$\large \left(AB \right)^{-1} = B^{-1} A^{-1} $$

<br>
This algebraic rule could be read as "the inverse of a product is equal to the product of the inverses." However, as it typically the case with matrices, the order of the matrices in the product matters! We can easily find examples where $ B^{-1} A^{-1} \neq A^{-1} B^{-1} $, and so we are not generally able to swap the order of the product on the right side of the equation above.

We will consider an example to demonstrates these facts.


<br>


### Instructions

Use the cell below to perform the following tasks:

1. Create lists-of-lists or arrays to represent the following two matrices:

$$
A = \begin{bmatrix}
7 & 3\\
5 & 2 \\
\end{bmatrix}
\quad\quad\quad
B = \begin{bmatrix}
3 & 8\\
2 & 5 \\
\end{bmatrix}
$$

2. Use tools from NumPy to calculate each of the following matrices, storing the results in Python variables.

$$
A^{-1} \quad\quad
B^{-1} \quad\quad
A B \quad\quad
(A B)^{-1} \quad\quad
B^{-1} A^{-1} \quad\quad
A^{-1} B^{-1}
$$





```python
A = np.array([[7, 3],
              [5, 2]])
B = np.array([[3, 8],
              [2, 5]])

A_inv = np.linalg.inv(A)
B_inv = np.linalg.inv(B)
AB = np.matmul(A, B)
AB_inv = np.linalg.inv(AB)
B_inv_A_inv = np.matmul(B_inv, A_inv)
A_inv_B_inv = np.matmul(A_inv, B_inv)
```

# Problem 2 - Magnitude or Length of Vectors

### Problem Background

The **length** or **magnitude** of a vector is a number that, in some way, represents the size of the vector. There is a nice geometric way of thinking about the length of a vector, which you can (optionally) learn more about in the following video from the rootmath YouTube channel: [Length of a 3D Vector](https://www.youtube.com/watch?v=MoHMw0ZO7bU)

We will now discuss how to calculate the magnitude of a vector. Consider the following N-dimensional vector:

$$\large v = [v_1, v_2, v_3, ..., v_N]  $$

<br>
The length or magnitude the vector $v$ is denoted by $\|v\|$ and is calculated using the formula shown below:

$$\large \|v\| = \sqrt{\sum_{i=1}^N v_i^2} $$


<br>

### Instructions

In the cell below, create a function named `magnitude()` with a single parameter named `v`. This parameter is intended to be a list or 1D array representing a vector.

The first line of your function should convert `v` to an array using the following code: `v = np.array(v)`

Your function should use tools from NumPy to calculate magnitude of `v`, which should then be returned by the function.


<br>

### Guidance

The steps in the process of calculating $\|v\|$ can be described as follows:

1. Square all of the elements of $v$.
2. Calculate the sum of the square of the elements of $v$.
3. Take the square root of the sum of the squares.

You can perform these steps with a single line of code or multiple lines of code.


```python
def magnititude(v):
  v = np.array(v)
  v_sq = v ** 2
  sum_v_sq = np.sum(v_sq)
  return sum_v_sq ** (1/2)
```

# Problem 3  - L-Norms

### Problem Background

The magnitude of a vector that was discussed in Problem 2 is a special case of something called an **L-norm**. Suppose that a vector $v$ is given as follows:

$$\large v = [v_1, v_2, v_3, ..., v_N]  $$

<br>

Let $L$ be a positive integer. The **L-norm** of the vector $v$ is denoted by $\|v\|_L$ and is calculated using the formula shown below:

$$\large \|v\|_L = \left(\sum_{i=1}^N |v_i|^L \right) ^{1/L} $$
<br>

The magnitude of a vector is a special case of the L-norm where $L=2$.

Versions of the L-norm (especially the 1-norm and 2-norm) are used in machine learning to apply **regularization techniques** that can be used to prevent models from over-fitting. You will likely encounter regularization in courses relating to machine learning and deep learning.

<br>

### Instructions

In the cell below, create a function named `lnorm()` with two parameters, `v` and `L`. The parameter `v` is intended to be a list or 1D array that represents a vector. The parameter `L` is intended to be a positive integer.

The first line of your function should convert `v` to an array using the following code: `v = np.array(v)`

Your function should use tools from NumPy to calculate the L-norm of `v`, which should then be returned by the function.

<br>

### Guidance

The steps in the process of calculating $\|v\|_L$ can be described as follows:

1. Find the absolute value of the elements of $v$ using `np.abs()`.
2. Raise each element of the array from step 1 to the power `L`.
3. Calculate the sum of the array created in step 2.  
4. Raise the sum from step 3 to the power `1/L`.

You can perform these steps with a single line of code or multiple lines of code. Pay careful attention to the order of operations and to your use of parentheses for grouping calculatations.


```python
def lnorm(v, L):
  v = np.array(v)
  v_abs_raised = np.abs(v) ** L
  return np.sum(v_abs_raised) ** (1/L)
```

# Problem 4  - Cosine Similarity

### Problem Background

The **cosine similarity** is a measure often used to indicate the similarity between two vectors. The cosine similarity for two vectors $x$ and $y$ is sometimes denoted as $S_C(x,y)$ and is calculated as shown below.

$$ S_C(x, y) =  \frac{x \cdot y}{ \|x\| \cdot \|y\| }$$

<br>

Geometrically speaking, the cosine similarity is equal to the cosine of the angle between the two vectors, and thus measures the degree to which the two vectors point in the same direction. It is used in machine learning and artificial intelligence to compare objects represented as vectors. Its usage is especially common in the field of Natural Language Processing.

<br>

### Instructions

In the cell below, create a function named `cosine_sim()` with two parameters, `x` and `y`. Both parameters are inteded to be lists or 1D arrays representing vectors. Your function should use tools from NumPy to calculate the cosine similarity of the two vectors. This value should be returned by the function.

<br>

### Guidance

Note that the numerator involves calculating a dot product. This can be accomplished using `np.dot()`. The denominator is equal to the product of the magnitudes of `x` and `y`. You can use your own `magnitude()` function from Problem 2 or the `np.linalg.norm()` function to calculate these magnitudes.



```python
def cosine_sim(x, y):
  return np.dot(x, y) / (magnititude(x) * magnititude(y))
```

# Problem 5 - Eigenvalues of an Inverse Matrix

### Problem Background

It can be shown that the eigenvalues of the inverse of a matrix $A$ are inverses of the eigenvalues of $A$ itself. Let's state this more formally. Suppose that $A$ is an invertible matrix and that $\lambda$ is an eigenvalue of $A$. Then $1/\lambda$ will be an eigenvalue of $A^{-1}$. We will demonstrate that fact in this problem.

<br>

### Instructions

Create a list-of-lists or an array to represent the following matrix:

$$
A = \begin{bmatrix}
8 & 3 & 0\\
-6 & 1 & 2 \\
6 & 3 & 2\\
\end{bmatrix}
$$

<br>

Use the `np.linalg.eig()` function to find the eigenvalues and eigenvectors for $A$. Calculate $A^{-1}$ and then find the eigenvalues and eigenvectors of $A^{-1}$.

Print both sets of eigenvalues. You do not need to print the eigenvectors. You should see that the eigenvalues of $A^{-1}$ are the reciprocals (or multiplicative inverses) of the eigenvalues of $A$.



```python
A = np.array([[8, 3, 0],
              [-6, 1, 2],
              [6, 3, 2]])

A_eigenvalues, A_eigenvectors = np.linalg.eig(A)

A_inv = np.linalg.inv(A)
A_inv_eigenvalues, A_inv_eigenvectors = np.linalg.eig(A_inv)

print(A_eigenvalues)
print(A_inv_eigenvalues)
```

# Problem 6 - Parameters for a Linear Regression Model

### Problem Background

In one of this week's lessons, we presented a matrix formula for calculating the optimal parameters for a linear regression model. A special case of that formula for a model with only a single feature is shown below.

$$ \large \hat \beta =
\begin{bmatrix}
\hat \beta_0 \\
\hat \beta_1 \\
\end{bmatrix} =
\left(X^T \cdot X \right ) ^{-1} \cdot X^T \cdot y
$$

<br>

### Instructions
The following three objects are defined in the cell below.
* A list named `x` that contains feature values for 10 observations.
* A list named `y` that contains target or label values for the same 10 observations.
* A 2D array named `X`. The first column of this array consists of ones and the second columns contains the 10 feature values. This array represents the matrix $X$ referenced in the formula above.

Use NumPy tools to calculate the parameter array $\beta$ by applying the calculations shown in the formula above. Store the final result in a variable named `beta`, and then print this array.

<br>

### Guidance

This calculation can be done in one line, but it might be easier to break in down into smaller calculations representing intermediate steps in the calculation. You can assign the results of the intermediate calculations to variables with names such as `step1`, `step2`, and so on.




```python
x = [32, 49, 38, 27, 46, 30, 13, 38, 11, 50]
y = [68, 89, 96, 44, 82, 54, 30, 72, 36, 83]
X = np.vstack([np.ones(10), x]).T

#-------------------------------
# Add your code here
#-------------------------------
step1 = np.matmul(X.T, X)
step2 = np.linalg.inv(step1)
step3 = np.matmul(step2, X.T)
beta = np.matmul(step3, y)
```

The cell below will create a scatter plot of the observations whose feature and label values are stored in `x` and `y`. It will also display the regression line corresponding to the values stored in your `beta` array. Run this cell as-is.

```python
import matplotlib.pyplot as plt
x_ep = np.array([np.min(x), np.max(x)])
y_ep = beta[0] + beta[1] * x_ep
plt.figure(figsize=[4,2])
plt.scatter(x, y, zorder=2)
plt.plot(x_ep, y_ep, color='darkorange', zorder=1)
plt.show()
```
