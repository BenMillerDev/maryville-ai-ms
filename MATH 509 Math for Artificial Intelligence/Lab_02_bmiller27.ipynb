# Feature Scaling

In this lab, we will discuss a data preprocessing task known as "feature scaling". This discussion provides context for Problem 1 and Problem 2.

In Week 1, we provided an introduction to **supervised learning**. Recall that the goal of a supervised learning problem is to create a **model** that can be used to predict the value of a **target** variable based on the values of one or more other **feature** variables.

The features in a supervised learning problem can reflect many different types of things and could be measured in any sort of unit imagineable. As a result, the ranges of values for features can (and often will) vary considerably from one feature to the next. This variability in the scale of the features can cause serious problems for certain types of machine learning algorithms. For many algorithms, it is useful (and sometimes necessary) to apply transformations to each feature so that the features are all on similar scales before using them to train a supervised learning model. This task is known as **feature scaling**.

In this lab, you will explore two common feature scaling techniques. In Problem 1, you will explore **min-max scaling** (also known as **normalization**). In Problem 2, you will explore **standard scaling**.

You will practice these techniques by scaling the values for a single feature. In each case, you will scale two sets of values for this feature. The first set of values is assumed to come from the **training set** of observations that will be used to train the model. The second set of values is assumed to come from new observations that were not included in the training set. These sets of feature values are provided in the arrays created in the cell below.

When applying feature scaling (or any other preprocessing technique), any parameters required for the transformation must be learned from the training set, regardless of what set they are being applied to.



```python
X_train = np.array(
    [29.8, 24.9, 30.7, 19.2, 31.8, 32.6, 34.4, 22.1,
     20.4, 19.1, 12.7, 13.7, 19.4, 19.6, 21.1, 17.1]
)

X_new = np.array([11.2, 22.8, 38.5, 25.2, 13.5, 25.5])
```

# Problem 1 - Min-Max Scaling

### Problem Background

We will now discuss **min-max scaling**. When applying min-max scaling to a feature, the training set values for that feature are scaled linearly to fall within the interval `[0, 1]`, with the minimum value scaled to 0 and the maximum value scaled to 1. Let's dig into the details.

Let $X$ refer to a feature in a dataset. Let $x_{\textrm{min}}$ be the smallest value of $X$ for observations in the training set and let $x_{\textrm{max}}$ be the largest value of $X$ for observations in the training set.

If $x_0$ is any observation of $X$, its scaled/normalized value is given by:

$$\large z_0 = \frac{x_0 - x_{\textrm{min}}}{x_{\textrm{max}} - x_{\textrm{min}}}$$

Note that the values $x_{\textrm{min}}$ and $x_{\textrm{max}}$ are taken from the **training set**. If and when scaling is to be applied to a new set of observations, those new observations must be scaled in the same way as the training set, using the same $x_{\textrm{min}}$ and $x_{\textrm{max}}$ values found from the training set.

When min-max scaling is applied to an observation from the training set, the scaled feature values will always be between 0 and 1 (inclusive). This rule does not have to hold when scaling is applied to observations outside of the training set.

<br>

# Problem 2 - Standard Scaling

We will now discuss **standard scaling**. When applying standard scaling to a feature, the training observations for that feature are scaled so that they have a mean of 0 and a standard deviation of 1.

Let $X$ refer to a feature in a dataset. Let $\bar x$ denote the mean of $X$ values for observations in the training set and let $s_X$ denote the standard deviation of $X$ for observations in the training set.

If $x_0$ is a value of $X$ for an observation, its scaled/normalized value is given by:

$$\large z_0 = \frac{x_0 - \bar x}{s_X}$$

As was the case for min-max scaling, the parameters for standard scaling ($\bar x$ and $s_X$) are calculated from the **training set**. When scaling a new set of observations, those new observations must be scaled in the same way as the training set, using the parameter values found from the training set.

<br>

An additional set of instructions for this problem is provided below.

# Problem 3 - Sum of Squared Error

### Problem Background

As discussed during Week 1, the **sum of squared errors** metric is often used for scoring regression models. We will review this concept now. Suppose that you are provided with $N$ true values for a target variable. We will denote these values as follows:

$$y_0, ~y_1, ~y_2, ~..., ~y_{N-1}$$

<br>

Suppose that you also have estimates for each of these values, denoted as follows:

$$\hat y_0, ~\hat y_1, ~\hat y_2, ~..., ~\hat y_{N-1}$$

<br>

We can calculate the error in each estimate as follows:

$$e_i = y_i - \hat y_i ~~\mathrm{for} ~~i=0,1,2,...,N-1$$

<br>

The sum of squared errors is calculated as follows:

$$SSE = \sum_{i=0}^{N-1} e_i^2$$

<br>

You used a loop to calculate the sum of squared errors in Lab 01. But the operations involved in this calculate are easily vectorized using NumPy, eliminating the need for a loop. We will explore this idea in this problem.

### Problem Background

There are many metrics that can be used to score a supervised learning model. One alternative to the sum of squared errors metric is the **root mean squared errors** (or **RMSE**) metric. Suppose that we have calculated error terms $e_1$, $e_2$, ..., $e_N$ as described in Problem 3. Then the RMSE metric can be calculated using the formula shown below.

$$\large \textrm{RMSE} = \sqrt{\frac{\sum_{i=1}^{N}e_i^2}{N}}$$

# Problem 6 - Multiclass Classification

### Problem Background

When a classification problem involves more than 2 classes, we refer to it as a **multiclass classification** problem. If a multiclass classification problem has $K$ classes, we often encode these classes as integers 0, 1, 2, ..., $K-1$.

Class probability estimates are also used for multiclass classification problems, but in that case it is not sufficient to provide one probability per estimate, as was the case with binary classification. Instead, we will typically provide one probability estimate per class, with the sum of these probability estimates equalling 1. We refer to this collection of probability estimates as a **probability distribution** over the possible classes.

If a multiclass classification model is applied to a collection of observations, the results are typically stored in a two-dimensional array. If there are $N$ observations and $K$ classes, then the array will have a shape of `(N, K)`.

When generating class predictions for an observation, if the classes are encoded as mentioned above, then we can simply apply the argmax operation to the probability distribution to determine the index of the largest probability.

<br>

### Instructions

An array of class probability estimates is provided in the cell below. The classification problem in quesiton has 4 classes, and there are 10 total observations. Add code to this cell to print the shape of the array `p`.

Then, in the cell below that, determine the class probability estimates by performing the following steps:

1. Create a variable `N` to store the number of rows of `p`. This can be done by using the `shape` attribute, or by passing `p` to the `len()` function. You should NOT simply hard-code in a value for `N`.

2. Loop over each row of `p`. For each iteration of the loop, do the following:

    b. Append the class prediction to a list named `mc_classes`.

3. When the loop finishes executing, print the list of class predictions.

```python
mc_classes = p.argmax(axis=1)
print(mc_classes)
```

# Problem 8 - Precision and Recall

### Problem Background

In this problem, we will explore alternate metrics that can be used to measure the performance of a binary classification model. For this discussion, we will assume that the classes are encoded as 0 and 1, with Class 0 referred to as the **negative class** and Class 1 referred to as the **positive class**. With that in mind, every observation can be placed into one of the following four groups:

* An observation is a **True Positive** if the model predicts that the observation is positive, and the observed class is actually positive.

* An observation is a **False Positive** if the model predicts that the observation is positive, but the observed class is actually negative.

* An observation is a **True Negative** if the model predicts that the observation is negative, and the observed class is actually negative.

* An observation is a **False Negative** if the model predicts that the observation is negative, but the observed class is actually positive.

Let `TP`, `FP`, `TN`, and `FN` refer to the number of true positives, false positives, true negatives, and false negatives, respectively. These counts can be used to calculate two metrics that can be used to provide a summary of the model's performance with respect to the positive class. These metrics are known as **precision** and **recall**. Formulas for these metrics are provided below.

$$\textrm{Precision} = \frac{TP}{TP + FP} = \frac{TP}{\textrm{Number of Positive Predictions}}$$

$$\textrm{Recall} = \frac{TP}{TP + FN} = \frac{TP}{\textrm{Number of Positive Observations}}$$

<br>

Precision can be thought of as an estimate of the probability that a positive prediction will actually be correct. Recall can be interpreted as an estimate of the probability that a positive observation will be correctly identified by the model.


<br>

### Instructions

In the cell below, calculate the number of true postives, false positives, true negatives, and false negatives for the arrays provided in Problem 7. You will need to use array comparisons, the Boolean operator `&`, and the `np.sum()` function. Print each of these values on separate lines.

In the cell after that, use the counts to calculate the precision and recall scores. Print each score, rounded to 4 decimal places.

```python
TP = np.sum((y_true == 1) & (y_pred == 1))
TN = np.sum((y_true == 0) & (y_pred == 0))
FP = np.sum((y_true == 0) & (y_pred == 1))
FN = np.sum((y_true == 1) & (y_pred == 0))

print(TP)
print(TN)
print(FP)
print(FN)
```

```python
precision = TP / (TP + FP)
recall = TP / (TP + FN)

print(round(precision, 4))
print(round(recall, 4))
```
