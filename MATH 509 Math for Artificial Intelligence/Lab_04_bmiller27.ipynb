# Problem 1 - Finding Roots

### Problem Background

In this problem, we will use SymPy to find the roots of a polynomial function.

<br>

### Instructions

**Part A.** In the cell below, create a SymPy variable named `x`. Then create an expression named `f` to represent the polynomial function shown below.

$$\large f(x) = 0.05 x^4 - 0.75 x^3 + 1.1 x^2 + 8 x + 35$$

<br>

After defining `f`, use the `display()` function to display it.

```python
x = sym.Symbol('x')
expression = lambda x : 0.05*x**4 - 0.75*x**3 + 1.1*x**2 + 8*x + 35
f = expression(x)

display(f)
```

**<font size=4>Additional Instructions</font>**

**Part B.** Use the `nroots()` function from SymPy to find the roots of the function $f(x)$. Store the results in a variable named `roots`. Then print the value of this variable.

You should see four roots. The first two will be real numbers, but the second two will be complex numbers that contain imaginary terms.


```python
roots = sym.nroots(f)
print(roots)
```

**<font size=4>Additional Instructions</font>**

**Part C.** Run the cell below as-is. If you have named your variables as instructed and if your `roots` variable is correct, then you should see a plot of the function $f(x)$ with the two real roots indicated using orange dots.

```python
x_vals = np.linspace(-5,13.5,100)
y_vals = [f.subs({x:x_}) for x_ in x_vals]
plt.figure(figsize=[6,3])
plt.axhline(0, c='k', linewidth=0.5)
plt.axvline(0, c='k', linewidth=0.5)
plt.plot(x_vals, y_vals)
plt.scatter(roots[:2], [0,0], c='darkorange', s=60, edgecolor='k', zorder=2)
#plt.yscale('symlog')
plt.show()
```

# Problem 5 - Local Extrema

### Problem Background

In this problem, we will use the first and second derivatives of $f(x)$ to find and classify the critical points of the function.

<br>

### Instructions

**Part A.** Use the `nroots()` function from SymPy to find the critical points of the function `f(x)` from Problem 1. Store the roots in a variable named `crit_pts`, and then print this variable.

```python
crit_pts = sym.nroots(df1)
print(crit_pts)
```

**<font size=4>Additional Instructions</font>**

**Part B.** Calculate the values of the function $f(x)$ and the second derivative $f''(x)$ at each of the critical points. Store the values of the function in a list named `f_values` and store the values of the second derivative in a list named `df2_values`.

After constructing the lists, print both lists using two different print statements.

```python
f_values = []
df2_values = []

for point in crit_pts:
  f_values.append(f.subs({x:point}))
  df2_values.append(df2.subs({x:point}))

print(f_values)
print(df2_values)
```

**<font size=4>Additional Instructions</font>**

**Part C.** Run the cell below as-is. If your lists have been constructed correctly and named as instructed, you should see output classify each critical point as either a local minimum or a local maximum. You should also see a plot of $f(x)$ with the local extrema indicated by orange dots.

```python
for cp, fx, df2x in zip(crit_pts, f_values, df2_values):
    if df2x < 0:
        print(f'{cp:.4f} is a local maximum. The value of the function here is {fx:.4f}.')
    if df2x > 0:
        print(f'{cp:.4f} is a local minimum. The value of the function here is {fx:.4f}.')

x_vals = np.linspace(-5,13.5,100)
y_vals = [f.subs({x:x_}) for x_ in x_vals]
plt.figure(figsize=[6,3])
plt.axhline(0, c='k', linewidth=0.5)
plt.axvline(0, c='k', linewidth=0.5)
plt.plot(x_vals, y_vals)
plt.scatter(crit_pts, f_values, c='darkorange', s=60, edgecolor='k', zorder=2)
plt.show()
```

# Problem 6 - Calculating the Mean

### Problem Background

Let $x_1, x_2, ..., x_N$ be a sample of numerical observations. The mean of such a sample can be calculated as follows:

$$\large \bar x = \frac{1}{N}\sum_{i=1}^N x_i$$

The mean of a sample has some interesting geometric properties. In particular, it is the unique value that minimize the total squared deviation with respect to the points in the sample. To state this explicitly, define the **sum of squared deviations** function, $SSD(x)$, as follows:

$$\large SSD(x) = \sum_{i=1}^N \left(x_i - x \right)^2$$

Then the mean $\bar x$ is the value of $x$ for which $SSD(x)$ is minimized. We will explore this aspect of the mean in this problem.

<br>

### Instructions

**Part A.** A sample of 10 values is provided in the cell below. Add code to that cell to perform the following steps:

1. Define a SymPy symbol named `x`.
2. Create an expresion named `SSD` to represent the function $SSD(x)$. This can be accomplished as follows:

   a. Initialize `SSD` to 0.

   b. Use a loop to calculate each term in the sum that defines $SSD(x)$. Add each term to the variable `SSD`.

3. Use `sym.simplify()` to simplify the expression for $SSD(x)$, storing the result back into the variable `SSD`.

4. Use `display()` to display the expression representing $SSD(x)$.



```python
sample = [24, 75, 74, 37, 59, 78, 73, 78, 86, 62]

#-----------------------------
# Add your code here.
#-----------------------------
x = sym.Symbol('x')

SSD = 0
for xi in sample:
  SSD += (xi - x) ** 2

SSD = sym.simplify(SSD)
display(SSD)
```

**<font size=4>Additional Instructions</font>**

**Part B.** Perform the following steps in the cell below.
1. Calculate the derivative of $SSD(x)$.
2. Use `nroots()` to find the critical points of $SSD(x)$. There should be only one.
3. Print the critical point found in the previous step.

```python
df_SSD = SSD.diff()
crit_pts_SSD = sym.nroots(df_SSD)
print(crit_pts_SSD)
```

**<font size=4>Additional Instructions</font>**

**Part C.** We will now verfiy our calculation. Use `np.mean()` to calculate the mean of the sample, printing the result.

```python
sample_mean = np.mean(sample)
print(sample_mean)
```

# Problem 7 - Linear Regression Loss

### Problem Background

In this problem, we will explore how calculus techniques can be used to find the optimal paramater values for a machine learning model. We will focus our attention on a simplified version of the linear regression model known as the **no-intercept linear regression** model. Assuming that our model has only one feature $x$, the no-intercept model can be written as follows:

$$\large \hat y = m\cdot x$$

The variable $m$ is parameter that defines the model. The goal during training is to find the value of $m$ that minimizes the **sum of squared errors** metric for some training set of $(x_i, y_i)$ feature/label pairs. The definition of $SSE(m)$ is provided below.

$$\large SSE(m) = \sum_{i=1}^N \left(y_i - \hat y_i \right)^2 = \sum_{i=1}^N \left(y_i - m\cdot x_i \right)^2  $$


<br>

### Instructions

**Part A.** Two lists named `x_train` and `y_train` are provided below. These lists contain the feature and label values for a training set that we will use to train a no-intercept linear regression model.

Add code to the cell to perform the following steps:

1. Define a SymPy symbol named `m`.
2. Create an expression named `SSE` to represent the function $SSE(n)$. This can be accomplished as follows:

   a. Initialize `SSE` to 0.

   b. Use a loop to calculate each term in the sum that defines $SSE(m)$. Add each term to the variable `SSE`.

3. Use `sym.simplify()` to simplify the expression for $SSE(x)$, storing the result back into the variable `SSE`.

4. Use `display()` to display the expression representing $SSE(x)$.



```python
x_train = [60, 55, 66, 58, 65, 17, 72, 88, 42, 69, 51, 93, 95, 80, 29, 66, 74, 12, 89, 54]
y_train = [105, 79, 83, 68, 100, 58, 109, 133, 66, 118, 104, 149, 144, 132, 33, 80, 106, 9, 135, 72]

#-----------------------------
# Add your code here.
#-----------------------------
m = sym.Symbol('m')
SSE = 0

for i in range(len(x_train)):
  SSE += (y_train[i] - (m *x_train[i])) ** 2

SSE = sym.simplify(SSE)
display(SSE)
```

**<font size=4>Additional Instructions</font>**

**Part B.** We will calculate the optimal slope $m$ for the model in Problem 8. First, we will calculate the value of $SSE(x)$ for some (arbitrary) slope $m =2$. Use the `subs()` method to plug 2 into $SSE(x)$ and print the result.  

```python
arbitrary_slope = SSE.subs({m:2})
print(arbitrary_slope)
```

**<font size=4>Additional Instructions</font>**

**Part C.** Run the cell below as-is. This will display a scatter plot of the training set along with a line representing the (non-optimal) model $\hat y = 2x$ that we scored in the cell above.

```python
x_vals = np.arange(0,100,2)
y_vals = 2 * x_vals
plt.figure(figsize=[4,2])
plt.scatter(x_train, y_train, zorder=2)
plt.plot(x_vals, y_vals, color='darkorange', zorder=1)
plt.show()
```

# Problem 8 - Minimizing Loss

### Problem Background

In this problem, we will find the value of the parameter $m$ that minimized the $SSE(m)$ loss from Problem 7.

<br>

### Instructions

**Part A.** In the cell below, calculate the derivative of $SSE(m)$. Use `display()` to display the derivative

Then use `nroots()` to find the critical points by calculating the roots of the derivative.  Print the list of critical points. You should see only a single critical point.




```python
df_SSE = SSE.diff()
display(df_SSE)

crit_pts_SSE = sym.nroots(df_SSE)
print(crit_pts_SSE)
```

**<font size=4>Additional Instructions</font>**

**Part B.** We will now calculate the minimum value for $SSE(m)$. The critical point found above is contained in a list. Use indexing to access that value and store it in a variable named `m0`. Then use `subs()` to plug the critical point into the expression for $SSE(m)$. Print the result.

```python
m0 = crit_pts_SSE[0]
minimum_value = SSE.subs({m:m0})
print(minimum_value)
```

**<font size=4>Additional Instructions</font>**

**Part C.** Run the cell below as-is. This will display a scatter plot of the training set along with a line representing the optimal model $\hat y = m\cdot x$.

```python
x_vals = np.arange(0,100,2)
y_vals = m0 * x_vals
plt.figure(figsize=[4,2])
plt.scatter(x_train, y_train, zorder=2)
plt.plot(x_vals, y_vals, color='darkorange', zorder=1)
plt.show()
```
