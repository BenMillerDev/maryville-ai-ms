# Problem 1 - Descriptive Statistics

Let $X$ be a discrete random variable with probability mass function shown below.


| $x$  | 1  | 2  | 4  | 8  | 16 | 32 | 64 |
|------|----|----|----|----|----|----|----|
|$p(x)$|0.08|0.13|0.30|0.24|0.16|0.06|0.03|


## Part 1.A
Use the cell below to perform the following tasks:
1. Create an array to store the values of $X$.
2. Create an array to store the probabilities provide by $p(x)$.
3. Use NumPy operations to calculate $E[X]$.
4. Use NumPy operations to calculate $\textrm{Var}[X]$.
5. Print $E[X]$ rounded to 2 decimal places.
6. Print $\textrm{Var}[X]$ rounded to 2 decimal places.

```python
x = np.array([1, 2, 4, 8, 16, 32, 64])
p = np.array([0.08, 0.13, 0.3, 0.24, 0.16, 0.06, 0.03])

# calculate Expected Value
EX = np.dot(x, p)
EX2 = np.dot(x**2, p)
# calculate Variance
VarX = EX2 - EX**2

print(np.round(EX, 2))
print(np.round(VarX, 2))
```

## Part 1.B

Define a new random variable $\large Y = \textrm{ln}(X)$. Use the cell below to perform the following tasks.

1. Create an array to store the possible values for $Y$.
2. Use NumPy operations to calculate $E[Y]$.
3. Use NumPy operations to calculate $\textrm{Var}[Y]$.
4. Print $E[Y]$ rounded to 2 decimal places.
5. Print $\textrm{Var}[Y]$ rounded to 2 decimal places.


```python
y = np.log(x)

# calculate Expected Value
EY = np.dot(y, p)
EY2 = np.dot(y**2, p)
# calculate Variance
VarY = EY2 - EY**2

print(np.round(EY, 2))
print(np.round(VarY, 2))
```

# Problem 3 - Maximum of 3 Dice

Suppose that three ten-sided dice are rolled. The faces of each dice are numbered using integers 1 - 10. Define a random variable $M$ to be equal to the largest value rolled. For example, if the rolls are 3, 7, 4, then $m=7$. We will explore the distribution of $M$ using simulation.  




## Part 3.A

We will start by creating a sample of observations of $M$ Use the cell below to complete the following tasks:

1. Use NumPy to set a seed of 1.
2. Use `np.random.choice()` to create a 2D array with size `(10_000, 3)`, with entries randomly sampled from the integers 1 - 10.

```python
values, counts = np.unique(M, return_counts=True)
props = counts / counts.sum()

print(props.round(2))
```

Widget is a term sometimes used to refer to a generic manufactured good. Suppose a factory produces widgets. They sell the widgets to retailers in batches of 1000. It is known that 1 in every 200 widgets is defective. In other words, an individual widget has a 0.005 probability of being defective. Let $D$ be a random variable representing the number of defective widgets in a batch of 1000. In this problem, you will use simulation to approximate a distribution for $D$.


## Part 4.A

Use the cell below to complete the following tasks:

# Problem 5 - Ride Share Service

A random variable is said to followin a **Poisson** distribution if its support consists of all non-negative integers and its PMF is given by the following formula for some fixed parameter value $\lambda$.

$$p(x) = \frac{e^{-\lambda} \lambda ^ x}{x!}$$

Let $R$ be a random variable that represents the number of requests recieved by a ride share service in a certain city during a specific one-hour period. Suppose that $R$ follows a Poission distribution with $\lambda = 50$. We will use simulation to explore the random variable $R$.

We will start by generating a sample of 10,000 observations of $R$. NumPy provides a function `np.random.poisson()` for generating samples of the Poission distribution. We use that function in the cell below to generate our sample for $R$. We also print the first 20 observations in the sample. Run this cell as-is.

```python
np.random.seed(1)
requests = np.random.poisson(50, size=10000)
print(requests[:20])
```

## Part 5.A

Use the sample provided to estimate the expected value of $E[R]$ and the standard deviation $SD[R]$ of the number of ride requests during an hour. Print both values rounded to 2 decimal places.


```python
expected = np.mean(requests)
std_dev = np.std(requests, ddof=1)

print(round(expected, 2))
print(round(std_dev, 2))
```

## Part 5.B

Use the sample provided to estimate the probabilities mentioned below. Print each probability rounded to 4 decimal places and in the the order listed.

* $P[R \leq 40]$

* $P[R \geq 50]$

```python
P1 = np.mean(requests <= 40)
P2 = np.mean(requests >= 50)

print(round(P1, 2))
print(round(P2, 2))
```

## Part 5.C

Use `np.percentile()` to calculate the 1st percentile and the 99th percentile of the number of rides requested in the simulated sample. Print both percentiles.

```python
percentile_1 = np.percentile(requests, 1)
percentile_99 = np.percentile(requests, 99)

print(percentile_1)
print(percentile_99)
```

Complete the following tasks in the cell below.

# Problem 6 - Selection in Genetic Algorithms

**Genetic algorithms** are a classical family of A.I. algorithms in which many possible candidate solutionsfor a problem are generated, and then selection, mutation, and reproduction processes are applied to "evolve" the solutions over time. We won't disucss all of the details here, we will take a look at how the selection process is performed.

When performing selection for a genetic algorithm, each candidate solution is assigned a **fitness score**. Each candidate is then assigned a probability of being selected. This probability is proportional to the candidate's fitness score. We then draw a sample of the desired size from the population of solutions. The sample is created according the to fitness-based probabilities and without replacement.

You will explore this selection process in this example. The code cell is used to create and print fitness scores for a population of 100 candidate solutions. The fitness scores are randomly assigned for the sake of this example, and range from 0 to 50. Run this cell as-is.

```python
np.random.seed(1)
fitness = np.random.choice(range(51), size=100)
print(fitness)
```

## Part 6.A

Calculate selection probabilities for each of the 100 candidate solutions. This is accomplished by dividing each fitness value by the total fitness for all 100 observations. Use NumPy operations rather than loops or list comprehensions.

Print the selection probabilities for the first 10 candidates, rounded to 4 decimal places.

```python
selection_prob = fitness / fitness.sum()

print(selection_prob[:10].round(4))
```

## Part 6.B

Perfrom the following tasks in the cell below:
1. Use NumPy to set a seed of 1.
2. Use `np.random.choice()` to sample 20 integer values ranging from 0 to 99. Perform the sampling without replacement and using the probabilities found in Part 6.B. This sample will represent the indices of observations to be selected by the algorithm.
3. Print the sample of indices. If everything was done correctly, the first value in the sample should be 45 and the last value should be 36.

```python
np.random.seed(1)
indices = np.random.choice(range(100), p=selection_prob, size=20, replace=False)
print(indices)
```

## Part 6.C

Select the fitness values of the selected individuals by passing the sample of indices to the `fitness` array. Print the fitness values of the selected individuals.

```python
selected_fitnesses = fitness[indices]
print(selected_fitnesses)
```

## Part 6.D

We will now confirm that this selection process did favor individuals with higher fitness scores. In the cell below, calculate and print the following values:

* The average fitness for all 100 solutions.

* The average fitness for the 20 selected solutions.

```python
mean_all = np.mean(fitness)
mean_selected = np.mean(selected_fitnesses)

print(mean_all)
print(mean_selected)
```
