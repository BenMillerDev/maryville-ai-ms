```python
import math
from collections import Counter
```

```python
def term_frequency(words):
    total_words = len(words)
    word_counts = Counter(words)

    tf_scores = {}
    for word, count in word_counts.items():
        tf_scores[word] = count / total_words

    return tf_scores
```

```python
def inverse_document_frequency(docs_words_dict):
    ## pass in a dictionary containing each document and its word list
    ## docs_words_dict = {"Dogs.txt" : [...], "Cats.txt": [...], "Birds.txt": [...]}
    total_docs = len(docs_words_dict)
    
    unique_words = set()
    for words_list in docs_words_dict.values():
        for word in words_list:
            unique_words.add(word)

    # Count how many docs contain each word
    idf_scores = {}
    for word in unique_words:
        docs_containing_word = 0
        for words_list in docs_words_dict.values():
            if word in words_list:
                docs_containing_word += 1
        
        idf_scores[word] = math.log(total_docs / docs_containing_word)

    return idf_scores
        
```

```python
def compute_tfidf(tf_scores, idf_scores):
    tfidf_scores = {}
    for word, tf in tf_scores.items():
        tfidf_scores[word] = tf * idf_scores[word]

    return tfidf_scores
```

```python
def print_top_scores(filename, tfidf):
    print(f"Top 5 TF-IDF words in '{filename}':")

    tfidf_copy = tfidf.copy()
    # find the 5 best tfidf scores
    for i in range(5):
        best_word = ""
        best_score = -1
        # loop through and find the best score
        for word, score in tfidf_copy.items():
            if score > best_score:
                best_score = score
                best_word = word
        # remove it so we can find the next best word in the next iteration
        del tfidf_copy[best_word]
        print(f"    {best_word}: {best_score:.4f}")
```

```python
with open("Dogs.txt") as file:
    dogs = file.read()
with open("Cats.txt") as file:
    cats = file.read()
with open("Birds.txt") as file:
    birds = file.read()


dogs_words = dogs.lower().split()
cats_words = cats.lower().split()
birds_words = birds.lower().split()

print(dogs, "\n")
print(cats, "\n")
print(birds, "\n")

docs_words_dict = {"Dogs.txt" : dogs_words, "Cats.txt" : cats_words, "Birds.txt" : birds_words}

# compute term frequency for each doc
tf_per_doc = {}
for filename, words_list in docs_words_dict.items():
    tf_per_doc[filename] = term_frequency(words_list)

# compute idf scores
idf_scores = inverse_document_frequency(docs_words_dict)

# compute TF-IDF for each doc
for filename, tf_scores in tf_per_doc.items():
    tfidf = compute_tfidf(tf_scores, idf_scores)
    print_top_scores(filename, tfidf)
```
