```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
path = "mpg.csv"
df = pd.read_csv(path)

# prepare the data for plotting
displ_year_df = df[["displ", "year"]].groupby("year")
lx = []
ln = []
# iterate over very groupby group and append their values as a list
# THIS IS A CRUCIAL STEP
for _, df_ in displ_year_df:
  lx.append(df["displ"].values.tolist())
  ln.append(list(set(df_["year"].values.tolist()))[0])

# instanciate the figure
fig = plt.figure(figsize = (8, 8))
ax = fig.add_subplot()

# blue and orange
colors = ["#0000FF", "#FF5C00"]

# plot the data
n, bins, patches = ax.hist(lx, bins = 30, stacked = True, density = False, color = colors)

# prettify the plot
ax.set_xticks(bins[::3])

ax.legend({class_:color for class_, color in zip(ln, colors)})

ax.set_title("Stacked histogram of engine displacement colored by year")

ax.set_ylabel("Frequency")
```

```python
PATH = "mpg.csv"
df = pd.read_csv(PATH)

plt.figure(figsize = (10, 10), dpi= 80)

sns.violinplot(x = "year",
               y = "hwy",
               data = df,
               density_norm='width',
               inner = 'quartile',
               hue = "year",
               legend = False
              )

# prettify the plot

# get the current figure
ax = plt.gca()
# get the xticks to iterate over
xticks = ax.get_xticks()

# iterate over every xtick and add a vertical line
# to separate different classes
for tick in xticks:
    ax.vlines(tick + 0.5, 0, np.max(df["hwy"]), color = "grey", alpha = .1)

# rotate the x and y ticks
ax.tick_params(axis = 'x', labelrotation = 90, labelsize = 12)
ax.tick_params(axis = 'y', labelsize = 12)

# add x and y label
ax.set_xlabel("Year", fontsize = 14)
ax.set_ylabel("HWY", fontsize = 14)

# set title
ax.set_title("Violin plot of highway mpg colored by year", fontsize = 14)
```
